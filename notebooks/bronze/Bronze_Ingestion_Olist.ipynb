```python
dbutils.widgets.text("src","")
src_value=dbutils.widgets.get("src")
print(src_value)
```

```python
df=spark.readStream.format("cloudFiles")\
    .option("cloudFiles.format","csv")\
    .option("header",True)\
    .option("inferSchema",True)\
    .option("cloudFiles.schemaLocation",f"/Volumes/workspace/bronze_olist/{src_value}/schema")\
    .option("cloudFiles.schemaEvolutionMode","rescue")\
    .option("path",f"/Volumes/workspace/olist_ecommerce/{src_value}")\
    .load()
```

```python
df.writeStream.format("delta").trigger(once=True)\
    .outputMode("append")\
    .option("checkpointLocation",f"/Volumes/workspace/bronze_olist/{src_value}/checkpoint")\
    .option("path",f"/Volumes/workspace/bronze_olist/{src_value}/data")\
    .start()
```
